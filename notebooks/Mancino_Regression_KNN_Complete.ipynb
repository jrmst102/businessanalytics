## 3. Define predictors and split by date
Use only information available before the selling day: store, category, weekday, and day of year. Store numbers are identifiers, so they are treated as categories. Day of year supplies a simple numeric time feature; it is not a full seasonality model.

The earliest 60% of dates form the training set, the next 20% form the validation set, and the latest 20% form the test set. Every observation from a given date stays in the same set. This asks whether patterns learned from earlier dates carry forward to later dates for the existing stores and categories.

Train on the training set, choose settings using validation results, and open the test set only after those choices are complete. This starter does not use lagged sales, promotions, or inventory data.


```python
df['store_id'] = df['store_id'].astype(str)
df['weekday'] = df['date'].dt.day_name()
df['day_of_year'] = df['date'].dt.dayofyear
CATEGORICAL = ['store_id', 'category', 'weekday']
NUMERIC = ['day_of_year']
FEATURES = CATEGORICAL + NUMERIC
TARGET = 'units_sold'

dates = np.sort(df['date'].unique())
train_end = dates[int(len(dates) * 0.60) - 1]
validation_end = dates[int(len(dates) * 0.80) - 1]
train = df.loc[df.date <= train_end].copy()
validation = df.loc[(df.date > train_end) & (df.date <= validation_end)].copy()
test = df.loc[df.date > validation_end].copy()
assert train.date.max() < validation.date.min() <= validation.date.max() < test.date.min()

X_train, y_train = train[FEATURES], train[TARGET]
X_val, y_val = validation[FEATURES], validation[TARGET]
X_test, y_test = test[FEATURES], test[TARGET]
display(pd.DataFrame([
    {'Set': name, 'Rows': len(part), 'First date': part.date.min().date(),
     'Last date': part.date.max().date()}
    for name, part in [('Training', train), ('Validation', validation), ('Test', test)]
]))
```

```python
# Explore training outcomes only; keep later outcomes for evaluation.
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].hist(y_train, bins=30, color='#356DA8', edgecolor='white')
axes[0].set(xlabel='Daily units per store/category', ylabel='Number of observations',
            title='Training sales distribution')
train.groupby('date').units_sold.sum().plot(ax=axes[1], color='#356DA8')
axes[1].set(xlabel='Date', ylabel='Total units across stores/categories',
            title='Training sales over time')
plt.tight_layout()
plt.show()
```

## 4. Prepare the KNN workflow
One-hot encoding turns categories into indicator columns. Scaling puts the numeric predictor on a comparable scale so day of year does not dominate the KNN distance calculation simply because its raw numbers are larger.

Each KNN configuration gets a fresh pipeline. Encoders and scalers are fitted on the training data only, then applied to validation data without refitting. Neighbors are selected using Euclidean distance in the transformed feature space.

| Metric | Interpretation | Preferred direction |
|---|---|---|
| MAE | Average absolute error, in units per store/category/day | Lower |
| RMSE | Error in units, with larger misses receiving more weight | Lower |
| R² | Performance relative to predicting the evaluation set's mean | Higher |

R² can be negative. It is not a percentage of predictions that are correct. Use validation MAE to select *k* for this exercise.


```python
def make_pipeline(estimator):
    preparation = ColumnTransformer([
        ('categories', OneHotEncoder(handle_unknown='ignore', sparse_output=False), CATEGORICAL),
        ('numeric', StandardScaler(), NUMERIC)
    ])
    return Pipeline([('prepare', preparation), ('model', estimator)])

def metrics(actual, predicted):
    return {'MAE': mean_absolute_error(actual, predicted),
            'RMSE': np.sqrt(mean_squared_error(actual, predicted)),
            'R2': r2_score(actual, predicted)}

# Rerunning this cell clears previous model runs.
models, results = {}, {}

def fit_and_evaluate(name, estimator):
    pipeline = make_pipeline(estimator)
    started = time.perf_counter()
    pipeline.fit(X_train, y_train)
    seconds = time.perf_counter() - started
    train_scores = metrics(y_train, pipeline.predict(X_train))
    val_scores = metrics(y_val, pipeline.predict(X_val))
    models[name] = pipeline
    results[name] = {'Model': name, 'Train MAE': train_scores['MAE'],
                     'Validation MAE': val_scores['MAE'],
                     'Validation RMSE': val_scores['RMSE'],
                     'Validation R2': val_scores['R2'], 'Fit seconds': seconds}
    display(pd.DataFrame([results[name]]).round(3))
    return pipeline
```

## 5. Establish a baseline
The baseline always predicts the training set's mean sales. A more complicated model must earn its place by improving on this simple rule. A later extension could use historical averages for each store, category, and weekday as a stronger baseline.


```python
baseline = fit_and_evaluate('Mean baseline', DummyRegressor(strategy='mean'))
```

## 6. K-nearest neighbors regression
KNN finds the *k* most similar training observations and averages their observed unit sales. With `weights='uniform'`, each neighbor receives equal weight. For example, if three neighbors sold 20, 25, and 30 units, the prediction would be 25 units.

The cell below fits *k* = 5, 15, and 50 using the same predictors and date splits. Smaller values follow local variation more closely; larger values smooth predictions. Keep other settings unchanged so the comparison isolates the effect of *k*.

Training predictions can include each observation as its own neighbor, making training error optimistic. Use the later validation period to choose *k*. KNN averages past outcomes and cannot extrapolate beyond the range of its training targets.


```python
# Compare the three KNN settings from the starter exercise.
K_VALUES = [5, 15, 50]
assert K_VALUES and all(isinstance(k, int) and 1 <= k <= len(X_train) for k in K_VALUES)

# Remove previous KNN runs when this cell is rerun; retain the baseline.
for name in list(models):
    if name.startswith('KNN ('):
        models.pop(name)
        results.pop(name)

for k in K_VALUES:
    name = f'KNN (k={k})'
    fit_and_evaluate(name, KNeighborsRegressor(
        n_neighbors=k, weights='uniform', metric='euclidean', n_jobs=-1))
```

## 7. Compare validation results and select k
Compare the three KNN settings with the baseline. A positive improvement percentage means lower MAE than the baseline; a negative value means worse performance. The code selects the KNN setting with the lowest validation MAE. If there is a tie, the first tied setting in `K_VALUES` is retained.

A wide gap between training and validation MAE can indicate overfitting or a change over time. Better training performance alone does not establish a better model. If KNN does not beat the baseline on validation data, retain the baseline as the preferred approach.

Before opening the test set, explain which *k* performed best, how much it improved on the baseline, and what an error of that size would mean for replenishment planning. If you change `K_VALUES`, rerun sections 6 and 7 before the final test.


```python
assert 'Mean baseline' in results, 'Run the baseline cell first.'
knn_names = [name for name in results if name.startswith('KNN (')]
assert knn_names, 'Run the KNN exercise first.'

comparison = pd.DataFrame(results.values()).sort_values(
    'Validation MAE', kind='stable').reset_index(drop=True)
baseline_mae = results['Mean baseline']['Validation MAE']
comparison['MAE improvement vs baseline (%)'] = (
    100 * (1 - comparison['Validation MAE'] / baseline_mae)
    if baseline_mae > 0 else np.nan)
display(comparison.round(3))

selected_name = min(knn_names, key=lambda name: results[name]['Validation MAE'])
selected_k = models[selected_name].named_steps['model'].n_neighbors
preferred_name = (selected_name if results[selected_name]['Validation MAE'] < baseline_mae
                  else 'Mean baseline')
print('Selected KNN setting:', selected_name)
print('Preferred approach using validation MAE:', preferred_name)

comparison.set_index('Model')['Validation MAE'].plot.barh(color='#356DA8')
plt.xlabel('Validation MAE: units per store/category/day')
plt.title('KNN settings compared with the mean-sales baseline')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## 8. Final test: open only after selection
The switch below is off by default. After selecting *k*, change it to `True`. The code refits the selected KNN configuration on training plus validation data and evaluates it on the untouched test period. It also refits the mean-sales baseline on the same data for comparison. The choice of *k* uses validation results only.

Do not tune settings after seeing these test results. If you do, the period becomes additional validation data and you need a new test period. Predictions are left unrounded for evaluation; fractional values represent expected units.

Use the output to complete this interpretation: “With *k* = ___, KNN's test MAE was ___ units per store/category/day, compared with ___ for the baseline. For replenishment planning, this means ___.” These predictions describe synthetic sales patterns; they do not establish causal effects of business decisions.


```python
RUN_FINAL_TEST = False        # Change to True only after selecting k in section 7.

if RUN_FINAL_TEST:
    assert 'selected_name' in globals(), 'Run section 7 to select k first.'
    assert selected_name in models, 'Rerun section 7 after changing KNN settings.'
    X_development = pd.concat([X_train, X_val])
    y_development = pd.concat([y_train, y_val])

    final_knn = clone(models[selected_name]).fit(X_development, y_development)
    test_predictions = final_knn.predict(X_test)
    final_baseline = DummyRegressor(strategy='mean').fit(X_development, y_development)
    baseline_predictions = final_baseline.predict(X_test)

    print('KNN setting selected using validation MAE:', selected_name)
    print('Approach preferred before opening the test set:', preferred_name)
    test_results = pd.DataFrame([
        {'Model': selected_name, **metrics(y_test, test_predictions)},
        {'Model': 'Refitted mean baseline', **metrics(y_test, baseline_predictions)}
    ])
    display(test_results.round(3))

    prediction_examples = test[['date', 'store_id', 'category', TARGET]].copy()
    prediction_examples['predicted_units'] = test_predictions
    prediction_examples['error_actual_minus_predicted'] = y_test.to_numpy() - test_predictions
    display(prediction_examples.head(10).round(2))

    plt.scatter(y_test, test_predictions, alpha=0.15, s=12, color='#356DA8')
    low = min(y_test.min(), test_predictions.min())
    high = max(y_test.max(), test_predictions.max())
    plt.plot([low, high], [low, high], color='#B64B44', linestyle='--')
    plt.xlabel('Actual units')
    plt.ylabel('Predicted units')
    plt.title(f'Final test: {selected_name}')
    plt.tight_layout()
    plt.show()
else:
    print('Test results remain unopened. Compare validation results and select k first.')
```
