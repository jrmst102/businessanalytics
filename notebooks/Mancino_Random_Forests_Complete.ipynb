## 3. Define predictors and split by date
Use only information available before the selling day: store, category, weekday, and day of year. Store numbers are identifiers, so they are treated as categories. Day of year supplies a simple numeric time feature; it is not a full seasonality model.

The earliest 60% of dates form the training set, the next 20% form the validation set, and the latest 20% form the test set. Every observation from a given date stays in the same set. This asks whether patterns learned from earlier dates carry forward to later dates for the existing stores and categories.

Train on the training set, choose settings using validation results, and open the test set only after those choices are complete. This notebook does not use lagged sales, promotions, or inventory data.


```python
df['store_id'] = df['store_id'].astype(str)
df['weekday'] = df['date'].dt.day_name()
df['day_of_year'] = df['date'].dt.dayofyear
CATEGORICAL = ['store_id', 'category', 'weekday']
NUMERIC = ['day_of_year']
FEATURES = CATEGORICAL + NUMERIC
TARGET = 'units_sold'

dates = np.sort(df['date'].unique())
train_end = dates[int(len(dates) * 0.60) - 1]
validation_end = dates[int(len(dates) * 0.80) - 1]
train = df.loc[df.date <= train_end].copy()
validation = df.loc[(df.date > train_end) & (df.date <= validation_end)].copy()
test = df.loc[df.date > validation_end].copy()
assert train.date.max() < validation.date.min() <= validation.date.max() < test.date.min()

X_train, y_train = train[FEATURES], train[TARGET]
X_val, y_val = validation[FEATURES], validation[TARGET]
X_test, y_test = test[FEATURES], test[TARGET]
display(pd.DataFrame([
    {'Set': name, 'Rows': len(part), 'First date': part.date.min().date(),
     'Last date': part.date.max().date()}
    for name, part in [('Training', train), ('Validation', validation), ('Test', test)]
]))
```

```python
# Explore training outcomes only; keep later outcomes for evaluation.
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].hist(y_train, bins=30, color='#356DA8', edgecolor='white')
axes[0].set(xlabel='Daily units per store/category', ylabel='Number of observations',
            title='Training sales distribution')
train.groupby('date').units_sold.sum().plot(ax=axes[1], color='#356DA8')
axes[1].set(xlabel='Date', ylabel='Total units across stores/categories',
            title='Training sales over time')
plt.tight_layout()
plt.show()
```

## 4. Prepare the random forest workflow
One-hot encoding turns categories into indicator columns. Random forests use threshold-based splits within individual trees, so numeric predictors do not need scaling. Day of year stays in its original units.

Each forest configuration gets a fresh pipeline. The encoder is fitted on training data only, then applied to validation data without refitting. Each tree uses squared-error splits and predicts the mean training sales in a terminal leaf. The forest averages its trees' predictions.

| Metric | Interpretation | Preferred direction |
|---|---|---|
| MAE | Average absolute error, in units per store/category/day | Lower |
| RMSE | Error in units, with larger misses receiving more weight | Lower |
| R² | Performance relative to predicting the evaluation set's mean | Higher |

R² can be negative. It is not a percentage of predictions that are correct. Use validation MAE to select forest size. The splitting criterion and the evaluation metric serve different purposes.


```python
def make_pipeline(estimator):
    preparation = ColumnTransformer([
        ('categories', OneHotEncoder(handle_unknown='ignore', sparse_output=False), CATEGORICAL),
        ('numeric', 'passthrough', NUMERIC)
    ])
    return Pipeline([('prepare', preparation), ('model', estimator)])

def metrics(actual, predicted):
    return {'MAE': mean_absolute_error(actual, predicted),
            'RMSE': np.sqrt(mean_squared_error(actual, predicted)),
            'R2': r2_score(actual, predicted)}

# Rerunning this cell clears previous model runs.
models, results = {}, {}

def fit_and_evaluate(name, estimator):
    pipeline = make_pipeline(estimator)
    started = time.perf_counter()
    pipeline.fit(X_train, y_train)
    seconds = time.perf_counter() - started
    train_scores = metrics(y_train, pipeline.predict(X_train))
    val_scores = metrics(y_val, pipeline.predict(X_val))
    models[name] = pipeline
    results[name] = {'Model': name, 'Train MAE': train_scores['MAE'],
                     'Validation MAE': val_scores['MAE'],
                     'Validation RMSE': val_scores['RMSE'],
                     'Validation R2': val_scores['R2'], 'Fit seconds': seconds}
    display(pd.DataFrame([results[name]]).round(3))
    return pipeline
```

## 5. Establish a baseline
The baseline always predicts the training set's mean sales. A more complicated model must earn its place by improving on this simple rule. A later extension could use historical averages for each store, category, and weekday as a stronger baseline.


```python
baseline = fit_and_evaluate('Mean baseline', DummyRegressor(strategy='mean'))
```

## 6. Random forest regression
A random forest combines many regression trees. Each tree is fitted on a bootstrap sample, drawn with replacement from the training observations. Candidate predictors are also randomly sampled at each split. These sources of variation produce different trees; averaging their predictions can reduce the instability of an individual tree.

The cell below compares 50, 100, and 200 trees. `n_estimators` controls the number of trees. Keep the other settings fixed so the comparison isolates forest size.

| Setting | Value | Purpose |
|---|---|---|
| `max_depth` | 12 | Limits the depth of each tree |
| `min_samples_leaf` | 30 | Requires at least 30 sampled training observations in a terminal leaf |
| `max_features` | 0.8 | Considers approximately 80% of the transformed feature columns at each split |
| `bootstrap` | True | Fits trees to samples drawn with replacement |
| `random_state` | 42 | Makes the random sampling reproducible in the same environment |

The transformed columns include the one-hot indicators, so `max_features` refers to those columns rather than just the four original predictors. More trees generally require more computation; they do not guarantee a lower validation error. Does any improvement justify the additional fitting time?


```python
N_TREES_VALUES = [50, 100, 200]
MAX_DEPTH = 12
MIN_LEAF = 30
MAX_FEATURES = 0.8
assert N_TREES_VALUES and all(isinstance(t, int) and t > 0 for t in N_TREES_VALUES)
assert isinstance(MAX_DEPTH, int) and MAX_DEPTH > 0
assert isinstance(MIN_LEAF, int) and MIN_LEAF > 0
assert 0 < MAX_FEATURES <= 1

# Remove previous forest runs when this cell is rerun; retain the baseline.
for name in list(models):
    if name.startswith('Forest ('):
        models.pop(name)
        results.pop(name)

for n_trees in N_TREES_VALUES:
    name = f'Forest (trees={n_trees})'
    fit_and_evaluate(name, RandomForestRegressor(
        n_estimators=n_trees, max_depth=MAX_DEPTH, min_samples_leaf=MIN_LEAF,
        max_features=MAX_FEATURES, bootstrap=True, criterion='squared_error',
        random_state=SEED, n_jobs=-1))
```

## 7. Compare validation results and select forest size
Compare each forest with the baseline. A positive improvement percentage means lower MAE than the baseline; a negative value means worse performance. The code selects the forest with the lowest validation MAE. If there is a tie, the smaller forest is preferred.

Compare training error, validation error, and fitting time. A wide gap between training and validation MAE can indicate overfitting or a change over time. Fitting time varies by runtime and does not include prediction time. If no forest beats the baseline on validation data, retain the baseline as the preferred approach.

Before opening the test set, explain which forest size performed best and whether the error changed materially as trees were added. If you change any settings, rerun sections 6–8 before the final test.


```python
assert 'Mean baseline' in results, 'Run the baseline cell first.'
forest_names = [name for name in results if name.startswith('Forest (')]
assert forest_names, 'Run the random forest exercise first.'

comparison = pd.DataFrame(results.values()).sort_values(
    'Validation MAE', kind='stable').reset_index(drop=True)
baseline_mae = results['Mean baseline']['Validation MAE']
comparison['MAE improvement vs baseline (%)'] = (
    100 * (1 - comparison['Validation MAE'] / baseline_mae)
    if baseline_mae > 0 else np.nan)
display(comparison.round(3))

selected_name = min(forest_names, key=lambda name: (
    results[name]['Validation MAE'], models[name].named_steps['model'].n_estimators))
selected_size = models[selected_name].named_steps['model'].n_estimators
preferred_name = (selected_name if results[selected_name]['Validation MAE'] < baseline_mae
                  else 'Mean baseline')
print('Selected random forest:', selected_name)
print('Preferred approach using validation MAE:', preferred_name)

comparison.set_index('Model')[['Train MAE', 'Validation MAE']].plot.barh(
    color=['#A7C4DE', '#356DA8'])
plt.xlabel('MAE: units per store/category/day')
plt.title('Training and validation errors by forest size')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## 8. Interpret the selected forest: feature importance
Permutation importance asks how much prediction error increases when one predictor's values are shuffled across validation observations. The selected forest remains fixed. Shuffling breaks the predictor's original relationship with the outcomes and other predictors.

The code shuffles each original predictor five times and reports the mean increase in validation MAE. The full pipeline handles encoding, so each categorical predictor is assessed as a whole rather than as separate indicator columns. This cell may take longer than fitting a forest because it makes repeated predictions.

A larger positive increase means the fitted model relies more on that predictor for this validation period. A value near zero means little measured contribution; a negative value means shuffling improved performance in that check. Error bars show variation across the five shuffles, not a confidence interval.

Which predictor contributes most to the selected forest's predictions? Would that information help explain the model to a store manager? Importance does not show whether a relationship is positive or negative, and it does not establish causation. Correlated predictors and unrealistic shuffled combinations can also affect the result. No test outcomes are used here.


```python
assert 'selected_name' in globals(), 'Run section 7 first.'
selected_pipeline = models[selected_name]

# With negative MAE scoring, a decrease in score is an increase in MAE.
importance = permutation_importance(
    selected_pipeline, X_val, y_val, scoring='neg_mean_absolute_error',
    n_repeats=5, random_state=SEED, n_jobs=1)
importance_table = pd.DataFrame({
    'Predictor': FEATURES,
    'Mean increase in MAE': importance.importances_mean,
    'SD across shuffles': importance.importances_std
}).sort_values('Mean increase in MAE', ascending=False)
print('Validation permutation importance for:', selected_name)
display(importance_table.round(3))

plt.figure(figsize=(9, 4))
plt.barh(importance_table['Predictor'], importance_table['Mean increase in MAE'],
         xerr=importance_table['SD across shuffles'], color='#356DA8', capsize=4)
plt.axvline(0, color='gray', linewidth=1)
plt.gca().invert_yaxis()
plt.xlabel('Increase in validation MAE after shuffling (units)')
plt.title('Which predictors does this forest rely on?')
plt.tight_layout()
plt.show()
```

## 9. Final test: open only after selection
The switch below is off by default. After selecting forest size, change it to `True`. The code refits the selected forest configuration on training plus validation data and evaluates it on the untouched test period. It also refits the mean-sales baseline on the same data for comparison. The forest-size choice uses validation results only.

Refitting can change the individual trees and their predictions. The importance analysis in section 8 describes the forest fitted on the training period, before this refit.

Do not tune settings after seeing these test results. If you do, the period becomes additional validation data and you need a new test period. Predictions are left unrounded for evaluation; fractional values represent expected units.

Use the output to complete this interpretation: “With ___ trees, the forest's test MAE was ___ units per store/category/day, compared with ___ for the baseline. For replenishment planning, this means ___.” These predictions describe synthetic sales patterns; they do not establish causal effects of business decisions.


```python
RUN_FINAL_TEST = False        # Change to True only after selecting forest size in section 7.

if RUN_FINAL_TEST:
    assert 'selected_name' in globals(), 'Run section 7 to select forest size first.'
    assert selected_name in models, 'Rerun section 7 after changing forest settings.'
    X_development = pd.concat([X_train, X_val])
    y_development = pd.concat([y_train, y_val])

    final_forest = clone(models[selected_name]).fit(X_development, y_development)
    test_predictions = final_forest.predict(X_test)
    final_baseline = DummyRegressor(strategy='mean').fit(X_development, y_development)
    baseline_predictions = final_baseline.predict(X_test)

    print('Forest setting selected using validation MAE:', selected_name)
    print('Approach preferred before opening the test set:', preferred_name)
    test_results = pd.DataFrame([
        {'Model': selected_name, **metrics(y_test, test_predictions)},
        {'Model': 'Refitted mean baseline', **metrics(y_test, baseline_predictions)}
    ])
    display(test_results.round(3))

    prediction_examples = test[['date', 'store_id', 'category', TARGET]].copy()
    prediction_examples['predicted_units'] = test_predictions
    prediction_examples['error_actual_minus_predicted'] = y_test.to_numpy() - test_predictions
    display(prediction_examples.head(10).round(2))

    plt.scatter(y_test, test_predictions, alpha=0.15, s=12, color='#356DA8')
    low = min(y_test.min(), test_predictions.min())
    high = max(y_test.max(), test_predictions.max())
    plt.plot([low, high], [low, high], color='#B64B44', linestyle='--')
    plt.xlabel('Actual units')
    plt.ylabel('Predicted units')
    plt.title(f'Final test: {selected_name}')
    plt.tight_layout()
    plt.show()
else:
    print('Test results remain unopened. Compare validation results and select forest size first.')
```
